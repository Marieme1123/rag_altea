### Q2 ✍ Pourquoi conserve-t-on le nom du fichier, la page et le service, et pas seulement le texte ?

Le texte seul ne dit pas d'où vient l'information. Avec la source, l'assistant peut :
1. **Citer ses sources** dans la réponse (« teletravail.pdf, page 1 ») : le salarié peut vérifier lui-même.
2. **Filtrer par service** (ne chercher que dans RH ou que dans IT) et préparer la confidentialité.

## partie Nettoyage 
### Q4 ✍ Un défaut qui reste et une règle pour le corriger

Avec le nettoyage de base (espaces, lignes vides, mots coupés), chaque page PDF garde un **en-tête** (« Altéa Services — Document interne — Diffusion restreinte »), un **numéro de page** et un **pied de page**. Ces lignes sont identiques sur toutes les pages : elles polluent chaque chunk.
**Règle ajoutée** : supprimer les lignes qui correspondent à l'en-tête et au pied de page, et les lignes qui ne contiennent qu'un nombre (`LIGNES_PARASITES`).
Défaut qui reste : les **tableaux** sont lus cellule par cellule (le plafond de repas et son montant sont sur deux lignes). Règle possible : un extracteur de tableaux (pdfplumber) ou regrouper les lignes deux par deux.

### Q5 : explication à la main
Le pas est taille − chevauchement = 4 − 1 = 3. Les chunks commencent aux positions 0, 3 et 6 : `abcd`, `defg`, `ghij`. Le dernier atteint la fin du texte, donc on s'arrête. Les lettres `d` et `g` sont répétées : c'est le chevauchement de 1.

### Q7 ✍ Risques et taille retenue
- **Chunk trop petit** : il perd son contexte (un montant sans le type de dépense, une phrase coupée) et on obtient beaucoup de chunks.
- **Chunk trop grand** : il mélange plusieurs sujets, son vecteur devient flou et la recherche moins précise.
- **Chevauchement** : il évite de perdre une information à la frontière entre deux chunks.
- **Taille retenue : 500 caractères, chevauchement 50.** Avec 200, les phrases sont souvent coupées ; avec 1000, un chunk mélange plusieurs rubriques.
